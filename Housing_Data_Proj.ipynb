```python
import os

# List the contents of the downloaded directory to find the CSV file
directory_contents = os.listdir(path)
print(f"Contents of the downloaded directory: {directory_contents}")

# Assuming there's one CSV file, find it
csv_file = [f for f in directory_contents if f.endswith('.csv')]

if csv_file:
    # Construct the full path to the CSV file
    csv_file_path = os.path.join(path, csv_file[0])
    print(f"Reading CSV file from: {csv_file_path}")
    df = pd.read_csv(csv_file_path)
    print('Successfully Loaded')
else:
    print("No CSV file found in the downloaded directory.")
```

```python
def clean_sqt(x):
  token=str(x).split("-")
  if len(token)==2:
    return (float(token[0])+ float(token[1]))/2
  try:
    return float(x)
  except:
    return None
```

```python
df['total_sqft'] = df['total_sqft'].apply(clean_sqt)
```
